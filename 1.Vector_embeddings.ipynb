These embeddings capture the semantic meaning of sentences, allowing us to compare them mathematically.

### Preparing the Data and Tools

In this cell, we perform two main actions:

1.  **Import Libraries**: We import `numpy` (as `np`) for numerical operations, especially useful when working with vectors and matrices, and `SentenceTransformer` from the `sentence_transformers` library, which is the core tool we'll use.
2.  **Define Sentences**: We create a Python list named `sentence` containing several strings. These are the sentences we want to convert into numerical representations (embeddings) to analyze their similarity.

```python
import numpy as np
from sentence_transformers import SentenceTransformer

sentence=["Machine learning is a subset of artificial intelligence",
"Deep learning uses nueral networks with many layers",
"natural language processing is popular for sentiment analysis and data science",
"azure provides cloud-based ai and ml services"]
```

### Generating Sentence Embeddings

Here, we take our sentences and transform them into **embeddings**:

1.  **Load Pre-trained Model**: `model = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')` loads a pre-trained neural network model. This specific model (`all-MiniLM-L6-v2`) is chosen for its efficiency and good performance in generating sentence embeddings.
2.  **Encode Sentences**: `embeddings = model.encode(sentence)` is the crucial step where the model processes each sentence in our `sentence` list and converts it into a high-dimensional numerical vector. These vectors are designed so that sentences with similar meanings are closer together in the vector space.
3.  **Display Embeddings**: `print(embeddings)` shows the generated numerical vectors. Each row in the output corresponds to an embedding for one of our input sentences.

```python
model = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')
embeddings=model.encode(sentence)
print (embeddings)
```

### Understanding the Shape of Embeddings

`embeddings.shape` tells us the dimensions of our `embeddings` array. The output `(4, 384)` means:

*   `4`: We have 4 sentences.
*   `384`: Each sentence has been converted into a vector with 384 dimensions. This number represents the complexity of the semantic information captured by the model for each sentence.

```python
embeddings.shape
```

### Inspecting a Single Embedding

`embeddings[1]` allows us to look at the numerical vector (embedding) for the second sentence in our `sentence` list (remember, Python uses 0-based indexing, so index 1 refers to the second item). This is just a demonstration to show the individual numerical values that make up one sentence's representation.

```python
embeddings[1]
```

### Querying for Similarities (Dot Product)

This section demonstrates how to find sentences similar to a given query using the generated embeddings. The core idea is to measure the **similarity** between the query's embedding and the embeddings of our existing sentences:

1.  **Input Query**: `query = input("enter your query: ")` prompts you to type a question or statement.
2.  **Encode Query**: `query_vec = model.encode([query])` converts your input query into its own 384-dimensional embedding, just like we did with the initial sentences.
3.  **Calculate Similarity (Dot Product)**: `scores = np.dot(embeddings, query_vec.T)` calculates the **dot product** between each of our stored sentence embeddings and the query embedding. The dot product is a mathematical operation that, in the context of embeddings, measures how 'aligned' two vectors are. A higher dot product typically indicates greater semantic similarity.
    *   `embeddings`: Our collection of sentence embeddings (4x384 matrix).
    *   `query_vec.T`: The transpose of our query embedding (384x1 vector). Transposing makes the dimensions compatible for dot product calculation.
4.  **Print Scores**: `print(scores)` displays the similarity scores. Each score corresponds to one of our original sentences, indicating its similarity to your query. The higher the score, the more similar the sentence is to the query.

```python
query= input("enter your query: ")
query_vec=model.encode([query])
scores=np.dot(embeddings, query_vec.T)
print(scores)

# Example Query and Output for this section:
# Query 1: Machine learning is AI?
# Output:
# [[0.79642236]
#  [0.34920692]
#  [0.3049997 ]
#  [0.5580896 ]]
# Query 2:Deep learning uses nueral networks with many layers
# Output:
# [[0.35919687]
#  [1.        ]
#  [0.12331748]
#  [0.3024181 ]]
# Query 3:What is the capital of India?
# Output:
# [[ 0.09631449]
# [-0.0469831 ]
# [-0.00284372]
# [ 0.03079253]]
```
